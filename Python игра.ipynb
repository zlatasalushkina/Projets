Мы предлагаем Вам почувствовать немного новогодней атмосферы с помощью небольшой игры. Ваш выбор влияет на исход игры, то есть на то получится ли приготовить традицонное оливье или все-таки нет, помните об этом и старайтесь следовать инструкциям.

```python
first_res = 'Ты молодец! Горошек найден! Новый год спасен! И даже не потребовалась помощь друзей! Ты и есть свой лучший помощник!'
second_res = 'Ура! Новый год с оливье и старыми друзьями! Разве это не идеально?!'
third_res = 'Эх...кажется, Новый год испорчен...придется праздновать без оливье...в следующем годы ты такого точно не допустишь!'
last_res = 'Кажется что-то пошло не так'

f_r = 0
s_r = 0
t_r = 0
l_r = 0

name = input('Привет, главный герой! Как тебя зовут?')

# действие первое
print('Наступило утро 31 декабря...')
a_1 = input('Готов ли ты встречать Новый год? (цифра):\n1 - Да, настроение отличное! Почти все салаты нарезаны! Уже жду праздник!\n2 - Что? Уже сегодня Новый год?\n')

print()

if a_1 == '1':
    print('Отлично!')
    print('Так...\nСеледка под шубой готова\nМандарины куплены\nПродукты для оливье... почти все в наличии\nОй... кажется, нет зеленого горошка\n')
    print('Что же делать?')
elif a_1 == '2':
    print('Разве нужно готовиться к празднику заранее?\nМандаринов у меня и так целый холодильник\nА что там с оливье?\nКажется, нет зеленого горошка для его приготовления')
    print('Что же делать?')

print()

# действие второе
a_2 = input('Твои действия (цифра):\n1 - Пора вставать с кровати и отправляться на поиски горошка\n2 - Наверное, у кого-то из друзей может быть зеленый горошек\n3 - Вечером дойду до ближайшего магазина, а сейчас пора смотреть новогодние фильмы!Новогоднее настроение само не появится\n')

print()
if a_2 == '1':
    print('\nНедалеко от дома есть Пятерочка, Магнит и Перекресток, начну с них')
elif a_2 == '2':
    print('\nРешено, наведаюсь к кому-нибудь из друзей, вопрос только к кому')
elif a_2 == '3':
    movie = input('\nКакой твой любимый новогодний фильм?\n')
    print(f'\nПришло время насладиться фильмом "{movie}"')

print()

# действие третье
print(f'\nВдруг раздался телефонный звонок!\nЗвонит твой друг детства, с которым вы не общались уже несколько лет.\nОн говорит: {name}, как дела? Не хочешь отпраздновать Новый год со мной и нашей общей подругой Настей?\n')

a_a_1 = input('Решите, хотите ли вы присоединиться к старым друзьям и отклониться от изначального плана на день (да или нет)\nесли вы ранее решили наведаться к друзьям выбирайте "да", ваши планы сбываются!\n')
# переменная с частью истории, будет использоваться не один раз
story1 = 'Друга зовут Ваня, и в школьные времена вы были действительно очень близки\nВаня уехал учиться во Францию, поэтому вы долгое время не виделись, но недавно он вернулся в Россию\nБыло бы очень интересно пообщаться и вспомнить школьные годы'
    
if a_a_1.lower() == 'да':
    print('\nУра спонтанным идеям!')
    print(story1)
    print('\nПриятно провести время с хорошими людьми! Может быть у них и горошек есть!')


if a_2 == '2' and a_a_1.lower() == 'нет':
    print('Какая-то ошибка...Ваня ведь твой друг')
    l_r += 1

else:
    if a_2 != '2' and a_a_1.lower() == 'нет':
        print()
        print(story1)
        print('\nНо планы на день менять как-то не хочется\nКонечно, провести время со старыми друзьями очень приятно, но можно сделать это в другой день\n')
    
    print()
    print('\nВнезапно раздается еще один телефонный звонок. Это мама, услышав, что для оливье не хватает горошка')
    
    from random import choice
    
    road = ['иди прямо', 'поверни направо', 'поверни налево', 'перейди дорогу']
    road_dict = {'иди прямо': '1', 'поверни направо': '2', 'поверни налево': '3', 'перейди дорогу': '4'}
    ro_ad = []
    
    print('Мама решила напомнить о небольшом продуктовом магазине, где он обычно есть и продиктовала маршрут:\n\nиди прямо,', end=' ')
    
    result_road = '1'
    
    for i in range(5):
        step = choice(road)
        ro_ad.append(step)
        result_road += road_dict[step]
    print(', '.join(ro_ad))
    
    print('\n\nСтоит записать маршрут, чтобы его не забыть (иди прямо - 1, поверни направо - 2, поверни налево - 3, перейди дорогу - 4): ')
    attempt = 2
    our_road = input()
    
    while True:
        if len(our_road) != 6:
            print('Что-то пошло не так. Мама говорила по-другому... действий было 6\n')
            our_road = input()
        elif attempt == 0 or our_road == result_road:
            break
        else:
            our_road = input('Хм...? Еще версия:\n')
            attempt -= 1
    print('\nОтлично!\n')
    

    if a_2 == '1' and a_a_1.lower() == 'нет':
        print('\nКажется, это совсем недалеко, можно попробовать сходить прямо сейчас, а если горошка там нет, то всегда остается вариант с Пятерочкой, Магнитом и Перекрестком\n')
        print('Дорога до магазина:')
        for symbol in our_road:
            for k, v in road_dict.items():
                if v == symbol:
                    print(k, end=' ')
        if our_road == result_road:
            print('\n\nВроде бы, вот нужный магазин')
        else:
            print('\n\nКажется, что-то пошло не так, нужно попросить помощи у прохожих\n')
            print('Высокий добродушный мужчина любезно подсказывает дорогу к магазину\n')
            print('--------------------В магазине-----------------------')
        print('\nТак... что-то полки с продуктами совсем пустые, видимо, все уже раскупили...\n')
        print('Стоит спросить у продавца, есть ли в наличии зеленый горошек\n')
        print('Ура! Продавец говорит, что горошек есть!')
        print('Праздник спасен! Пора готовить оливье')
        f_r += 1
        
    
    elif a_a_1.lower() == 'да':
            # действие четвертое
            # дом Вани
        print('\nМаршрут есть, но надеюсь, он не пригодится, пора идти к Ване\n')
        print('--------------------В доме Вани-----------------------')
        products = ['картофель', 'огурцы', 'помидоры', 'чай', 'колбаса', 'консервы', 'сахар', 'яйца', 'лук', 'соль', 'подсолнечное масло', 'приправы']
        products_needed = []
    
        while len(products_needed) != 6:
            item = choice(products)
            if item not in products_needed:
                products_needed.append(item)
        
        places = {'холодильник': ['картофель', 'огурцы', 'колбаса', 'яйца', 'лук'],
                  'верхняя тумбочка': ['сахар', 'соль', 'приправы'],
                  'нижняя тумбочка': ['консервы', 'подсолнечное масло']}
        
        print('Ваня:\n- Помоги, пожалуйста, с продуктами. Сможешь найти их на кухне и достать на стол?')
        print('Ты:\n- Конечно, что потребуется?')
        print(f'Ваня:\n- Для праздничного ужина нам нужны:\n{(", ").join(products_needed)}')
        print('Ты:\n- Ясно. Где все это хранится?')
        
        places_needed = {}

        for pr in products_needed:
            for k, v in places.items():
                if pr in v:
                    if k in places_needed:
                        places_needed[k].append(pr)
                    else:
                        places_needed[k] = [pr]
    
        print('\nИтак...')
    
        for k, v in places_needed.items():
            if len(v) > 1:
                print(f'{k}: {(", ").join(v[0: -1])} и {v[-1]}')
            else:
                print(f'{k}: {v[0]}')
    
        print('\nПродукты найдены!\n')
        print('Ты: Ваня, у меня с собой есть продукты для оливье, не хватает только зеленого горошка, у тебя есть?\nВаня: Конечно, он в кладовке\nТы: Ура! Праздник спасен! Тогда я схожу за ним!')

        pr_set = set(products_needed)
        peas_set = {'горошек'}
        pr_on_table = pr_set | peas_set
        print('\nВсе нужные продукты лежат на столе\nТак, что у нас здесь: ')
        print(*pr_on_table, sep = ', ')
        print('\nПраздник спасен! Пора готовить оливье и другие блюда')
        s_r += 1
        
    
    elif a_2 == '3' and a_a_1.lower() == 'нет':
        print(f'Ага... маршрут есть, проблема почти решена, вечером схожу в магазин, а пока меня ждет отдых и {movie}') 
        print()
        print('......много времени спустя.....')
        print()
        print(f'......Фильм "{movie}" был просто чудесен. Не зря ты его так любишь!\n\nНо уже совсем свечерело\nКажется, уже 9 вечера и все магазины закрыты\nВсе пропало, кажется, этот Новый год пройдет без оливье...')
        t_r += 1


with open('results.txt', 'w', encoding='utf-8') as outfile:
    if l_r == 1:
        print(f'Исход игры: {third_res}', file = outfile)
    elif f_r == 1:
        print(f'Исход игры: {first_res}', file = outfile)
    elif s_r == 1:
        print(f'Исход игры: {second_res}', file = outfile)
    elif t_r == 1:
        print(f'Исход игры: {third_res}', file = outfile)
```
